```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Dataset/statefarm/driver_imgs_list.csv')

# Count the number of images in each class
class_counts = df['classname'].value_counts()

# Create a bar plot
plt.figure(figsize=(10, 6))
class_counts.plot(kind='bar')
plt.title('Class Distribution of Statefarm Dataset')
plt.xlabel('Class Name')
plt.ylabel('Number of Images')
plt.show()
```

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

# Paths from your run
pred_path = Path("evaluations/clip_lora/statefarm_eval/clip_statefarm_shared7_from_samdd_lora/predictions.jsonl")
label_map_path = Path("Dataset/statefarm/statefarm_label_map_shared7.json")

# Load predictions
records = [json.loads(line) for line in pred_path.open()]
df = pd.DataFrame(records)

# Preserve the canonical class order from the label map
label_map = json.load(label_map_path.open())
class_order = list(label_map.keys())  # ['c0', 'c6', 'c4', 'c2', 'c3', 'c1', 'c9']

# Confusion matrix
cm = confusion_matrix(df["target"], df["prediction"], labels=class_order)
cm_df = pd.DataFrame(cm, index=class_order, columns=class_order)

plt.figure(figsize=(7, 6))
sns.heatmap(cm_df, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicted class")
plt.ylabel("True class")
plt.title("StateFarm shared-7 confusion matrix\n(SAM-DD LoRA zero-shot)")
plt.show()

# Text report for quick per-class precision/recall/F1
print(classification_report(df["target"], df["prediction"], labels=class_order, target_names=class_order))
```
